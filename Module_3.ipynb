```python
from datetime import datetime


class Account:
    def __init__(self, account_number, name, balance=0):
        self.account_number = account_number
        self.name = name
        self.__balance = balance
        self.transactions = []

        self.transactions.append(
            f"{datetime.now():%Y-%m-%d %H:%M:%S} | "
            f"Account created | Balance: ‡{balance:.2f}"
        )

    def deposit(self, amount):
        if amount <= 0:
            print("Please enter a positive amount.")
            return

        self.__balance += amount
        self.transactions.append(
            f"{datetime.now():%Y-%m-%d %H:%M:%S} | "
            f"Deposit: ‡{amount:.2f} | Balance: ‡{self.__balance:.2f}"
        )
        print("Deposit successful!")

    def withdraw(self, amount):
        if amount <= 0:
            print("Please enter a positive amount.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            self.transactions.append(
                f"{datetime.now():%Y-%m-%d %H:%M:%S} | "
                f"Withdrawal: ‡{amount:.2f} | Balance: ‡{self.__balance:.2f}"
            )
            print("Withdrawal successful!")

    def show_balance(self):
        print(f"Account Holder: {self.name}")
        print(f"Account Number: {self.account_number}")
        print(f"Current Balance: ‡{self.__balance:.2f}")

    def show_history(self):
        print("\n--- Transaction History ---")
        for transaction in self.transactions:
            print(transaction)


def main():
    accounts = {}

    while True:
        print("\n===== PYTHON BANKING SYSTEM =====")
        print("1. Create Account")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. View Account Balance")
        print("5. View Transaction History")
        print("6. Exit")

        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                account_number = input("Enter account number: ").strip()

                if not account_number:
                    print("Account number cannot be empty.")
                    continue

                if account_number in accounts:
                    print("Account number already exists.")
                    continue

                name = input("Enter account holder name: ").strip()
                if not name:
                    print("Name cannot be empty.")
                    continue

                balance = float(input("Enter initial deposit: ‡"))

                if balance < 0:
                    print("Initial deposit cannot be negative.")
                    continue

                accounts[account_number] = Account(
                    account_number, name, balance
                )
                print("Account created successfully!")

            elif choice in ("2", "3", "4", "5"):
                account_number = input("Enter account number: ").strip()
                account = accounts.get(account_number)

                if account is None:
                    print("Account not found. Please create an account first.")
                    continue

                if choice == "2":
                    amount = float(input("Enter deposit amount: ‡"))
                    account.deposit(amount)

                elif choice == "3":
                    amount = float(input("Enter withdrawal amount: ‡"))
                    account.withdraw(amount)

                elif choice == "4":
                    account.show_balance()

                elif choice == "5":
                    account.show_history()

            elif choice == "6":
                print("Thank you for using the Banking System!")
                break

            else:
                print("Invalid choice. Please select 1-6.")

        except ValueError:
            print("Invalid input. Please enter a valid number.")


if __name__ == "__main__":
    main()
```
